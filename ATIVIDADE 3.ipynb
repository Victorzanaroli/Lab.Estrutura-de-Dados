```python
from stack import Stack

def is_balanced(expression):
    """
    Verifica se a expressão possui parênteses balanceados.
    Usa a pilha implementada em stack.py.
    """
    pilha = Stack()

    #------- COLOQUE SEU CÓDIGO AQUI -------
    # Dicionário para mapear os pares de fechamento com suas respectivas aberturas
    pares = {')': '(', ']': '[', '}': '{'}
    
    for char in expression:
        # Se for um caractere de abertura, empilha
        if char in pares.values():
            pilha.push(char)
            
        # Se for um caractere de fechamento
        elif char in pares.keys():
            # Se a pilha estiver vazia ou o topo não corresponder à abertura correta, é inválido
            if pilha.is_empty() or pilha.pop() != pares[char]:
                return False
                
    # No final, a expressão só estará balanceada se a pilha estiver vazia
    return pilha.is_empty()
    #---------------------------------------


# Alguns Testes
print(is_balanced("[{}(2+2)]{}")) # Esperado True
print(is_balanced("[{}(2+2))]{}")) # Esperado False
print(is_balanced("[{}])")) # Esperado False

print(" ")
print(is_balanced(""))                         # Expressão vazia → True
print(is_balanced("()"))                       # Um par de parênteses simples → True
print(is_balanced("({[]})"))                   # Vários pares aninhados → True
print(is_balanced("{[()()]()}"))               # Aninhamento complexo correto → True
print(is_balanced("if (a[0] > b) { return c; }")) # Parênteses misturados com código → True

print(" ")
print(is_balanced("("))                        # Apenas abertura → False
print(is_balanced("]"))                        # Apenas fechamento → False
print(is_balanced("({[)]}"))                   # Ordem incorreta → False
print(is_balanced("{[(()]}"))                  # Um fechamento errado no meio → False
print(is_balanced("((()))]"))                  # Um fechamento extra → False
```

```python
from stack import Stack

if __name__ == "__main__":
    main_stack = Stack()
    min_stack = Stack()

    def push_aux(data):
        """
        Função para empilhar na pilha principal e atualizar a pilha de mínimos quando necessário.
        """
        # Sempre adicionamos o dado na pilha principal
        main_stack.push(data)
        
        # Se a pilha de mínimos estiver vazia OU o dado atual for menor ou igual
        # ao mínimo atual (topo da min_stack), empilhamos na min_stack também.
        # O "<=" é importante para lidar com valores mínimos duplicados.
        if min_stack.is_empty() or data <= min_stack.peek():
            min_stack.push(data)

    def pop_aux():
        """
        Função para desempilhar da pilha principal e atualizar a pilha de mínimos quando necessário.
        """
        if main_stack.is_empty():
            raise IndexError("pop from empty stack")
            
        # Removemos o topo da pilha principal
        removido = main_stack.pop()
        
        # Se o valor removido for exatamente o mínimo atual, 
        # nós o removemos da pilha de mínimos também.
        if removido == min_stack.peek():
            min_stack.pop()
            
        return removido

    def get_min():
        """
        Função para retornar o mínimo atual.
        """
        # O mínimo atual sempre será o topo da min_stack
        if min_stack.is_empty():
            raise IndexError("A pilha está vazia")
            
        return min_stack.peek()

    # Testes
    print("\nEmpilhando: 5, 3, 7, 2, 8")
    push_aux(5)
    print(f"Min atual: {get_min()}")

    push_aux(3)
    print(f"Min atual: {get_min()}")

    push_aux(7)
    print(f"Min atual: {get_min()}")

    push_aux(2)
    print(f"Min atual: {get_min()}")

    push_aux(8)
    print(f"Min atual: {get_min()}")

    print("\nDesempilhando e mostrando o mínimo:")
    pop_aux() # Remove 8, min continua 2
    print(f"Min atual: {get_min()}")

    pop_aux() # Remove 2, min volta a ser 3
    print(f"Min atual: {get_min()}")

    pop_aux() # Remove 7, min continua 3
    print(f"Min atual: {get_min()}")

    pop_aux() # Remove 3, min volta a ser 5
    print(f"Min atual: {get_min()}")

    pop_aux() # Remove 5, pilha fica vazia
    try:
        print(get_min())
    except IndexError as e:
        print(f"Erro esperado: {e}")
```
