# RESOLUÇÃO

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class SingleLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0

    def append(self, data):
        new_node = Node(data)

        # Caso lista vazia
        if self.head is None:
            self.head = new_node
            self.tail = new_node
        else:
            # Liga o último nó ao novo nó
            self.tail.next = new_node
            self.tail = new_node

        self.size += 1

    def insert(self, index, data):
        # Verificação de índice inválido
        if index < 0 or index > self.size:
            raise IndexError("Índice fora do intervalo")

        new_node = Node(data)

        # Inserção no início
        if index == 0:
            new_node.next = self.head
            self.head = new_node

            # Se a lista estava vazia
            if self.size == 0:
                self.tail = new_node

        else:
            # Percorrer até o nó anterior ao índice
            prev = self.head
            for _ in range(index - 1):
                prev = prev.next

            # Inserção no meio ou final
            new_node.next = prev.next
            prev.next = new_node

            # Se inseriu no final, atualiza tail
            if new_node.next is None:
                self.tail = new_node

        self.size += 1

    def __str__(self):
        elements = []
        trav = self.head
        while trav:
            elements.append(str(trav.data))
            trav = trav.next
        return " -> ".join(elements)


# Teste
linked_list = SingleLinkedList()
linked_list.append(5)
linked_list.append(23)
linked_list.append(7)
linked_list.append(13)

print("Lista original:")
print(linked_list)

linked_list.insert(2, 11)

print("\nLista após inserir 11 na terceira posição:")
print(linked_list)
```
